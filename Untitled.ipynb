```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
     "Date": [
        "2025-01-02","2025-01-05","2025-01-08","2025-01-10","2025-01-15",
        "2025-01-18","2025-01-22","2025-01-25","2025-01-28","2025-02-02",
        "2025-02-05","2025-02-08","2025-02-10","2025-02-12","2025-02-15",
        "2025-02-18","2025-02-20","2025-02-22","2025-02-25","2025-02-28"
    ],
    
    "Product Name":[
        "laptop","iphone","tablet","ipad","gaming pod",
        "iphone","laptop","mouse","headsets","keyboard",
        "ipad","iphone","laptop","iphone","monitor",
        "laptop","iphone","mouse","headsets","monitor"
    ],
    
    "Units Sold":[
        "10","20","40","5","17",
        "20","30","22","11","25",
        "16","33","80","6","21",
        "24","52","18","56","15"
    ],
    
    "Revenue":[5000,4000,2400,1200,7000,5000,2000,1800,3000,4500,
               1600,2100,8000,3600,2500,2700,10000,4200,3000,3200],
    "Region":[
        "North","South","East","West","North",
        "South","East","West","North","South",
        "East","West","North","South","East",
        "West","North","South","East","West"
    ],
    "Discount Offered (%)":[10,5,15,0,8,10,5,20,0,12,18,5,10,0,8,15,5,10,0,12],
    "Salesperson":[
        "Alice","Bob","Charlie","David","Alice",
        "Bob","Charlie","David","Alice","Bob",
        "Charlie","David","Alice","Bob","Charlie",
        "David","Alice","Bob","Charlie","David"
    ]
}

df = pd.DataFrame(data)
df
```

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Revenue After Discount"] = df["Revenue"] * (1 - df["Discount Offered (%)"]/100)

df["Performance"] = np.where(df["Revenue"] >= 5000, "High", "Low")

print("\nSales Dataset")
print(df)

```

```python
print("\n1. Top 3 Revenue Transactions")
print(df.nlargest(3, "Revenue"))

```

```python
print("\n2. Total Units Sold Per Product")
print(df.groupby("Product Name")["Units Sold"].sum())

```

```python
print("\n3. Total Revenue After Discount")
print(df["Revenue After Discount"].sum())
```

```python
print("\n4. Transaction with Highest Discount")
max_discount = df.loc[df["Discount Offered (%)"].idxmax()]
print(max_discount[["Product Name","Discount Offered (%)","Revenue After Discount"]])
```

```python
print("\n5. Highest Total Revenue by Salesperson")
sales_rev = df.groupby("Salesperson")["Revenue"].sum()
print(sales_rev)
print("Top:", sales_rev.idxmax())
```

```python
print("\n6. Average Discount by Salesperson")
print(df.groupby("Salesperson")["Discount Offered (%)"].mean())
```

```python
print("\n7. Revenue by Region")
print(df.groupby("Region")["Revenue"].sum())
```

```python
print("\n8. Alice's Best Region")
alice = df[df["Salesperson"]=="Alice"]
print(alice.groupby("Region")["Revenue"].sum())
```

```python
print("\n10. High Performance Transactions")
print((df["Performance"]=="High").sum())
```

```python
print("\n11. North Region, No Discount")
north = df[(df["Region"]=="North") &
           (df["Discount Offered (%)"]==0)]
print(north.groupby("Salesperson")["Units Sold"].sum())
```

```python
print("\n13. Average Revenue After Discount")
print(df.groupby(["Salesperson","Region"])["Revenue After Discount"].mean())

```

```python
print("\n14. Cumulative Revenue")
df = df.sort_values(["Salesperson","Date"])
df["Cumulative Revenue"] = df.groupby("Salesperson")["Revenue"].cumsum()
print(df[["Date","Salesperson","Revenue","Cumulative Revenue"]])
```

```python
print("\n15. Top Two Transactions Per Salesperson")
df["Rank"] = df.groupby("Salesperson")["Revenue"]\
               .rank(method="dense", ascending=False)
print(df[df["Rank"]<=2][["Salesperson","Revenue","Rank"]])
```

```python
print("\n16. Product-wise Cumulative Revenue")
df = df.sort_values(["Product Name","Date"])
df["Product Cumulative Revenue"] = \
    df.groupby("Product Name")["Revenue"].cumsum()
print(df[["Date","Product Name","Revenue",
          "Product Cumulative Revenue"]])
```

```python
print("\n17. Revenue Comparison")
comparison = df.groupby("Product Name")[["Revenue",
                                         "Revenue After Discount"]].mean()
print(comparison)
```

```python
print("\n18. Weighted Average Discount")

weighted_discount = df.groupby("Salesperson").apply(
    lambda x: np.average(
        x["Discount Offered (%)"],
        weights=x["Revenue"]
    ),
    include_groups=False
)

print(weighted_discount)

```
